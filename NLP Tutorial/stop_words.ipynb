```python
# Exercise
# From a given text return the count of stop words
text = """Every morning, before the sun had fully risen over the hills, an old fisherman would walk down to the edge of the lake where his small wooden boat was tied. He had been doing this for as long as anyone in the village could remember, and although he was no longer as strong as he used to be, he refused to give up the routine that had defined most of his life. Some of the younger fishermen often asked him why he did not simply retire and let others take over, but he would only smile and say that the lake was the one place where he felt truly at peace.
One day, while he was preparing his nets, a young boy from the village approached him and asked if he could come along. The old man hesitated at first, because he was not used to having company on his trips, but there was something about the boy's curiosity that reminded him of himself many years ago. So he agreed, and together they pushed the boat out into the calm, glassy water.
As they drifted further from the shore, the boy asked a lot of questions — about the fish, about the weather, and about why the old man had never left the village even though he could have gone anywhere he wanted. The fisherman explained that it was not because he lacked the opportunity, but because he had never found anything that felt more important than this lake, this boat, and these quiet mornings. The boy did not fully understand what he meant, but he nodded anyway, because there was something comforting in the way the old man spoke.
By the time they returned to the shore, the sun was high in the sky, and the village had begun to wake up. The old man handed the boy a small fish and told him that if he wanted to learn how to fish properly, he should come back tomorrow, and the day after that, and every day until he no longer needed to ask why. The boy smiled and promised that he would.
"""

doc = nlp(text)

stop_words = [token for token in doc if token.is_stop]

print(f"Length of Stop Words: {len(stop_words)}")

# Find the percentage of stop words compared to total words in the text
total_words_count = len(doc)
total_stop_words_count = len(stop_words)

percentage_stop_words = (total_stop_words_count / total_words_count) * 100
print(f"Percentage of Stop words presented in the given text: {percentage_stop_words} %")
```

```python
# From a given text, print the frequency and most frequently used words
text = """Every morning, before the sun had fully risen over the hills, an old fisherman would walk down to the edge of the lake where his small wooden boat was tied. He had been doing this for as long as anyone in the village could remember, and although he was no longer as strong as he used to be, he refused to give up the routine that had defined most of his life. Some of the younger fishermen often asked him why he did not simply retire and let others take over, but he would only smile and say that the lake was the one place where he felt truly at peace.
One day, while he was preparing his nets, a young boy from the village approached him and asked if he could come along. The old man hesitated at first, because he was not used to having company on his trips, but there was something about the boy's curiosity that reminded him of himself many years ago. So he agreed, and together they pushed the boat out into the calm, glassy water.
As they drifted further from the shore, the boy asked a lot of questions — about the fish, about the weather, and about why the old man had never left the village even though he could have gone anywhere he wanted. The fisherman explained that it was not because he lacked the opportunity, but because he had never found anything that felt more important than this lake, this boat, and these quiet mornings. The boy did not fully understand what he meant, but he nodded anyway, because there was something comforting in the way the old man spoke.
By the time they returned to the shore, the sun was high in the sky, and the village had begun to wake up. The old man handed the boy a small fish and told him that if he wanted to learn how to fish properly, he should come back tomorrow, and the day after that, and every day until he no longer needed to ask why. The boy smiled and promised that he would.
"""

doc = nlp(text)

remaining_tokens = []

for token in doc:
    if not token.is_stop or not token.is_punct:
        remaining_tokens.append(token.text)

token_frequency = {}
for token in remaining_tokens:
    if token != "\n" and token != " ":
        if token not in token_frequency:
            token_frequency[token] = 1
        else:
            token_frequency[token] += 1

max_frequent_token = max(token_frequency.keys(), key=(lambda key: token_frequency[key]))

print(f"Maximum frequency word: {max_frequent_token}")
```
