```python
"""Step 1: Literal Matches & Character Classes (The Baseline)"""
"""
Crucial NLP Shortcuts to Memorize:
\w: matches any word character
\W: matches any non-word character
\s: matches any whitespace character
\d: matches any digit character
[^...]: matches any character not in the brackets
"""

# Code Example

import re

text = "User data: 1 active account, 0 pending items."

# Find all digits
digits = re.findall(r"\d", text)
print(f"Digits found: {digits}")

# Find all lower-case vowels
vowels = re.findall(r"[aeiou]", text)
print(f"Vowels found: {vowels}")

# Find all non-word characters
non_word_chars = re.findall(r"\W", text)
print(f"Non-word characters found: {non_word_chars}")

# Find all non-space characters
print(f"Non-space characters found: {re.findall(r"\S", text)}")

# Find all non-digit characters
non_digit_chars = re.findall(r"\D", text)
print(f"Non-digit characters found: {non_digit_chars}")

# Find any character that is not in the brackets
not_in_brackets = re.findall(r"[^...]", text)
print(f"Characters not in brackets found: {not_in_brackets}")
```

```python
"""Step 2: Quantifiers (Matching Varying Lengths)"""
"""
Crucial NLP Shortcuts to Memorize:
*: matches zero or more times
+: matches one or more times
?: matches zero or one times
{m}: matches exactly m times
{m,}: matches m or more times
{m,n}: matches between m and n times
"""

# Code Exapmple
import re

text = "The patient weighed 74kg in 2024 and 78kg in 2026."

# Extract all chunks of 1 or more word characters
words = re.findall(r"\w+", text)
print(f"Tokens: {words}")

# Extract only 4 digits years
years = re.findall(r"\d{4}", text)
print(f"Years: {years}")

# Extract only 2 to 3 digits weights ends with "kg"
weights = re.findall(r"\d{2,3}kg", text)
print(f"Weights: {weights}")

text = """
http://example.com
https://google.com
"""

# Use of ? to match zero or one time
print(re.findall(r"https?", text))

# Use of + to match one or more times
print(re.findall(r"http+", text))

# Use of * to match zero or more times
print(re.findall(r"https*", text))

text = """
Alice bought 25 apples.
Bob bought 300 oranges.
"""

# Extract all words
print(re.findall(r"\w+", text))

# Extract all numbers
print(re.findall(r"\d+", text))
```

```python
"""Step 3: Boundaries (Isolating Complete Words)"""
"""
Crucial NLP Shortcuts to Memorize:
^: matches the start of a string
$: matches the end of a string
\b: matches a word boundary
"""

# Code Example
import re

text = "AI training causes pain. The AI system is online."

# Without Boundaries
accidental_matches = re.findall(r"ai", text, re.IGNORECASE)
print(f"Without boundaries: {accidental_matches}")

# With boundaries: only catches the isolated word "AI"
exact_matches = re.findall(r"\bai\b", text, re.IGNORECASE)
print(f"With boundaries: {exact_matches}")
```

```python
"""Step 4: Text Cleaning & Substitution (The NLP Pipeline Core)"""

# Code Example

import re

raw_tweet = "Breaking News! The NLP pipeline is working... perfectly. #DataScience"

# Clean the tweet
cleaned_tweet = re.sub(r"[^\w\s]", "", raw_tweet)
print(f"Cleaned tweet: {cleaned_tweet}")
```

```python
"""Step 5: Groups and Alternation (Advanced Extraction)"""
"""
Crucial NLP Shortcuts to Memorize:
|: Acts as an OR operator
(): Creates a group to isolate and capture substrings
"""

# Code Example

import re

financial_text = "The item costs $45.00, while the shipping fee is €5.50."

price_pattern = r"(\$|€)(\d+\.\d{2})"

matches = re.findall(price_pattern, financial_text)
print(f"Extracted (Currency, Amount): {matches}")
```

```python
"""Extracting Words Properly"""

text_with_special_characters = "Python is a programming language that It's easy to learn."

# Use of [A-Za-z]+(?:`[A-Za-z]+)?
words = re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?", text_with_special_characters)
print(f"Words: {words}")

text = """
don't
e-mail
state-of-the-art
well-known
"""
# Use of [A-Za-z]+(?:[-`][A-Za-z]+)*
tokens = re.findall(
    r"[A-Za-z]+(?:[-'][A-Za-z]+)*",
    text
)
print(f"Tokens: {tokens}")
```

```python
import re

"""Sentence Extraction"""
text_with_multiple_sentences = """
The patient weighed 74kg in 2024 and 78kg in 2026.
Python is a programming language that It's easy to learn.
Will you learn Python?
"""

sentences = re.findall(r"[^.!?]+", text_with_multiple_sentences)
print(f"Sentences: {sentences}")

sentences = re.findall(r"[^.!?]+[.!?]", text_with_multiple_sentences)
print(f"Sentences: {sentences}")
```

```python
import spacy # type: ignore

nlp = spacy.load("en_core_web_sm")

text = ("When Sebastian Thrun started working on self-driving cars at "
        "Google in 2007, few people outside of the company took him "
        "seriously. “I can tell you very senior CEOs of major American "
        "car companies would shake my hand and turn away because I wasn’t "
        "worth talking to,” said Thrun, in an interview with Recode earlier "
        "this week.")

doc = nlp(text)

print(f"Noun phrases: {[chunk.text for chunk in doc.noun_chunks]}")
print(f"Verbs: {[token.lemma_ for token in doc if token.pos_ == 'VERB']}")

# Find named entities, phrases and concepts
for entity in doc.ents:
    print(f"{entity.text}, {entity.label_}")
```
