```python
import pandas as pd


df = pd.read_csv("spam.csv")
df.head()
```

```python
df["Category"].value_counts()
```

```python
df["Spam"] = df["Category"].apply(lambda x: 1 if x == "spam" else 0)
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df["Message"], df["Spam"], test_size=0.25, random_state=0)
```

```python
print(f"X_train size: {X_train.shape}")
print(f"X_test size: {X_test.shape}")
print(f"y_train size: {y_train.shape}")
print(f"y_train size: {y_test.shape}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train.values)
```

```python
X_train_vectorized.shape
```

```python
X_train_vectorized.toarray()
```

```python
vectorizer.get_feature_names_out()[-1]
```

```python
vectorizer.vocabulary_
```

```python
import numpy as np


np.where(X_train_vectorized.toarray()[0] != 0)
```

```python
# Build machine learning model
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB()
model.fit(X_train_vectorized, y_train)
```

```python
X_test_vectorizer = vectorizer.transform(X_test)
y_prediction = model.predict(X_test_vectorizer)
y_prediction
```

```python
from sklearn.metrics import accuracy_score, classification_report

print((f"Accuracy: {accuracy_score(y_test, y_prediction)}"))
print(f"Classification Report: \n {classification_report(y_test, y_prediction)}")
```

```python
emails = [
    "Hey mohan, can we get together to watch footbal game tomorrow?",
    "Upto 20% discount on parking, exclusive offer just for you. Dont miss this reward!"
]

emails_count_vector = vectorizer.transform(emails)
model.predict(emails_count_vector)
```

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("naive_bayes_classifier", MultinomialNB()), 
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
emails = [
    "Hey mohan, can we get together to watch footbal game tomorrow?",
    "Upto 20% discount on parking, exclusive offer just for you. Dont miss this reward!"
]

pipeline.predict(emails)
```

```python
print((f"Accuracy: {accuracy_score(y_test, pipeline.predict(X_test))}"))
print(f"Classification Report: \n {classification_report(y_test, pipeline.predict(X_test))}")
```

```python
"""Exercise"""
import pandas as pd
# Load the dataset
df = pd.read_csv("IMDB Dataset.csv")
df.head()
```

```python
# Print the shape of the dataset
df.shape
```

```python
df["sentiment"].value_counts()
```

```python
df["sentiment_boolean"] = df["sentiment"].apply(lambda x: 1 if x == "positive" else 0)
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df["review"], df["sentiment_boolean"], test_size=0.2, random_state=0)
```

```python
print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train.values)
X_test_vectorizer = vectorizer.transform(X_test.values)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=50, criterion="entropy", random_state=0)
model.fit(X_train_vectorized, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

print(f"Accuracy Score: {accuracy_score(y_test, model.predict(X_test_vectorizer))}")
print(f"Classification repost: \n {classification_report(y_test, model.predict(X_test_vectorizer))}")
```

```python
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("random_forest_classifier", RandomForestClassifier(n_estimators=50, criterion="entropy", random_state=0))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
print(f"Accuracy Score: {accuracy_score(y_test, pipeline.predict(X_test))}")
print(f"Classification repost: \n {classification_report(y_test, pipeline.predict(X_test))}")
```

```python
from sklearn.neighbors import KNeighborsClassifier

pipeline = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("knn_classifier", KNeighborsClassifier(n_neighbors=10, metric="euclidean"))
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
print(f"Accuracy Score: {accuracy_score(y_test, pipeline.predict(X_test))}")
print(f"Classification repost: \n {classification_report(y_test, pipeline.predict(X_test))}")
```

```python
pipeline = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("naive_bayes_classifier", MultinomialNB())
])

pipeline.fit(X_train, y_train)

print(f"Accuracy Score: {accuracy_score(y_test, pipeline.predict(X_test))}")
print(f"Classification repost: \n {classification_report(y_test, pipeline.predict(X_test))}")
```
