```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = ("When Sebastian Thrun started working on self-driving cars at "
        "Google in 2007, few people outside of the company took him "
        "seriously. “I can tell you very senior CEOs of major American "
        "car companies would shake my hand and turn away because I wasn’t "
        "worth talking to,” said Thrun, in an interview with Recode earlier "
        "this week.")

doc = nlp(text)

print(f"Noun phrases: {[chunk.text for chunk in doc.noun_chunks]}")
print(f"Verbs: {[token.lemma_ for token in doc if token.pos_ == 'VERB']}")

# Find named entities, phrases and concepts
for entity in doc.ents:
    print(f"{entity.text}, {entity.label_}")
```

```python
from spacy.symbols import ORTH


doc = nlp("Gimme a sentence.")

nlp.tokenizer.add_special_case(
    "Gimme", 
    [{ORTH: "Gim"}, {ORTH: "me"},]
)

for token in doc:
    print(token.text, " | ", token.pos_, " | ", token.lemma_, " | ", token.dep_, " | ", token.orth_)
```

```python
doc = nlp("Artificial Intelligence is a new field of computer science.")

print(nlp.pipe_names)

if "sentencizer" not in nlp.pipe_names:
    print("No 'sentencizer' component. Adding it now.")
    nlp.add_pipe("sentencizer")

print(nlp.pipe_names)

for sentence in doc.sents:
    print(sentence)
    for word in sentence:
        print(word)
```

```python
text = """
Look for data to help you address the question. Governments are good
sources because data from public research is often freely available. Good
places to start include http://www.data.gov/, and http://www.science.
gov/, and in the United Kingdom, http://data.gov.uk/.
Two of my favorite data sets are the General Social Survey at http://www3.norc.org/gss+website/, 
and the European Social Survey at http://www.europeansocialsurvey.org/.
"""

# Extract all URL from the given text
doc = nlp(text)

for sentence in doc.sents:
    for word in sentence:
        if word.like_url:
            print(word)
```

```python
text = "I have 2 $, and 10 €."
doc = nlp(text)

# Extract all money with the currency symbol
for sentence in doc.sents:
    for word in sentence:
        if word.like_num:
            print(word)
        if word.is_currency:
            print(word)

for sentence in doc.sents:
    for word in sentence:
        if word.like_num and doc[word.i + 1].is_currency:
            print(word, doc[word.i + 1])
```

```python
import nltk
import spacy


from nltk import PorterStemmer

stemmer = PorterStemmer()

words = [
    "eating",
    "eats",
    "eat",
    "eaten",
    "eater",
    "adjustable",
    "adjust",
    "adjusts",
    "adjusted",
    "adjusting", 
    "ability", 
    "able", 
    "meeting", 
    "meet", 
    "met"
]

for word in words:
    print(word, " | ", stemmer.stem(word))
```

```python
nlp = spacy.load("en_core_web_sm")

text = " ".join(words)

doc = nlp(text)

for token in doc:
    print(token.text, " | ", token.lemma_, " | ", token.lemma)
```

```python
print(nlp.pipe_names)

attribute_ruler = nlp.get_pipe("attribute_ruler")

attribute_ruler.add(
    [
        [
            {
                "TEXT": "Bro"
            }
        ], 
        [
            {
                "TEXT": "Brah"
            }
        ]
    ], 
    {
        "LEMMA": "Brother"
    }
)

text = "Bro Brah"

doc = nlp(text)

for token in doc:
    print(token.text, " | ", token.lemma_, " | ", token.lemma)
```

```python
# Stemming using nltk
words = ['running', 'painting', 'walking', 'dressing', 'likely', 'children', 'whom', 'good', 'ate', 'fishing']

from nltk import PorterStemmer

stemmer = PorterStemmer()

for word in words:
    print(word, " | ", stemmer.stem(word))
```

```python
# Lemmatization using spacy
doc = nlp("running painting walking dressing likely children who good ate fishing")

for token in doc:
    print(token.text, " | ", token.lemma_, " | ", token.lemma)
```

```python
# Using stemming in nltk
text = "Artificial Intelligence is a new field of computer science. It is a rapidly growing field."

base_words = []
word_tokens = nltk.word_tokenize(text)

from nltk import PorterStemmer

stemmer = PorterStemmer()

for word_token in word_tokens:
    print(word_token, " | ", stemmer.stem(word_token))
    base_words.append(stemmer.stem(word_token))

print(" ".join(base_words))
```

```python
text = "Artificial Intelligence is a new field of computer science. It is a rapidly growing field."

doc = nlp(text)
base_words = []

for token in doc:
    print(token.text, " | ", token.lemma_, " | ", token.lemma)
    base_words.append(token.lemma_)

print(" ".join(base_words))
```
