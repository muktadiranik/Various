```python
"""Name Entity Recognition (NER)"""
import spacy


nlp = spacy.load("en_core_web_sm")

nlp.pipe_names
```

```python
doc = nlp("This is a chat fine-tuned version of google-bert/bert-base-uncased, designed to perform Named Entity Recognition on a text sentence imput.")

for entity in doc.ents:
    print(entity.text, " | ", entity.label_, " | ", spacy.explain(entity.label_))
```

```python
from spacy import displacy


displacy.render(doc, style="ent")
```

```python
print(nlp.pipe_labels["ner"])
```

```python
nlp.pipe_labels["ner"]

for entity in doc.ents:
    for label in nlp.pipe_labels["ner"]:
        if label == entity.label_:
            print(entity.text, " | ", entity.label_, " | ", spacy.explain(entity.label_))
```

```python
from spacy.tokens import Span

span_1 = Span(doc, 9, 10, label="ORG")

doc.set_ents([span_1], default="unmodified")
displacy.render(doc, style="ent")

for entity in doc.ents:
    print(entity.text, " | ", entity.label_, " | ", spacy.explain(entity.label_))
```

```python
text = """
Maria spent three weeks backpacking across Southeast Asia last spring. She started in Bangkok, Thailand, where she wandered through the narrow lanes near the Chao Phraya River before catching a flight to Hanoi, Vietnam. From there, she took an overnight train north to Sapa, a mountain town near the border with China, known for its terraced rice fields in the Hoang Lien Son range.
"""

doc = nlp(text)

# Extract all geographical location names
for entity in doc.ents:
    if entity.label_ == "GPE":
        print(entity.text, " | ", entity.label_, " | ", spacy.explain(entity.label_))

print(f"Count of GPE: {len([entity for entity in doc.ents if entity.label_ == 'GPE'])}")
```

```python
text = """
The city council announced on March 3, 2024 that construction on the new library would begin the following month. Officials confirmed the groundbreaking ceremony would take place on April 15, with an expected completion date sometime in late 2025. This isn't the first major project for the district — the community center down the street was finished back in 2019, just before the pandemic disrupted similar plans across the country. Residents have been asked to submit feedback by the end of next week, and a follow-up town hall is scheduled for 9:00 AM on Monday, May 6th.
"""

doc = nlp(text)

# Extract all date entities
for entity in doc.ents:
    if entity.label_ == "DATE":
        print(entity.text, " | ", entity.label_, " | ", spacy.explain(entity.label_))

print(f"Count of DATE: {len([entity for entity in doc.ents if entity.label_ == 'DATE'])}")
```
